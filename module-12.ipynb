```python
source = 0

cap = cv2.VideoCapture(source)

win_name = "Camera Preview"
net = cv2.dnn.readNetFromCaffe("deploy.prototxt", "res10_300x300_ssd_iter_140000_fp16.caffemodel")

in_width = 300
in_height = 300
mean = [104, 177, 123]

conf_threshold = 0.7
last_updated_time = 0.0
x = 0
cv2.namedWindow(win_name, cv2.WINDOW_NORMAL)
while cv2.waitKey(1) != 27:
    has_frame , frame = cap.read()
    if not has_frame:
        break

    frame = cv2.flip(frame, 1)
    frame_height = frame.shape[0]
    frame_width = frame.shape[1]
    blob = cv2.dnn.blobFromImage(frame, 1.0, (in_width, in_height), mean, swapRB = False, crop = False)

    net.setInput(blob)
    detections = net.forward()

    # The shape of detections is [1,1,N,7]

    for i in range(detections.shape[2]):
        confidence = detections[0,0,i,2]
        if confidence > conf_threshold:
            x_top_left = int(detections[0,0,i,3]*frame_width)
            y_top_left = int(detections[0,0,i,4]*frame_height)
            x_bottom_right = int(detections[0,0,i,5]*frame_width)
            y_bottom_right = int(detections[0,0,i,6]*frame_height)

            cv2.rectangle(frame, (x_top_left, y_top_left), (x_bottom_right, y_bottom_right), (0,0,255), thickness = 2, lineType = cv2.LINE_AA)
            label = "Confidence:%.4f"%confidence
            label_size, base_line = cv2.getTextSize(label, cv2.FONT_HERSHEY_SIMPLEX, 0.5, 1)
            cv2.rectangle(frame, (x_top_left, y_top_left-label_size[1]),(x_top_left+label_size[0], y_top_left), (255,255,255), -1)
            cv2.putText(frame, label, (x_top_left, y_top_left), cv2.FONT_HERSHEY_SIMPLEX, 0.5,(0,0,0),1)
    if last_updated_time == 0.0:
        last_updated_time = cv2.getTickCount()
    elapse_secs = (cv2.getTickCount() - last_updated_time)/ cv2.getTickFrequency()

    if elapse_secs > 0.5:
        t, _ = net.getPerfProfile()
        x = (t*1000/(cv2.getTickFrequency()))
        last_updated_time = cv2.getTickCount()
    label = "Inference time: %.2f ms" % x
    cv2.putText(frame, label, (0, 15), cv2.FONT_HERSHEY_SIMPLEX, 0.5, (0,255,0))
    cv2.imshow(win_name, frame)

cap.release()
cv2.destroyWindow(win_name)

    
```
