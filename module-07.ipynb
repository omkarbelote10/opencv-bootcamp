```python
import cv2
import sys
import numpy as np


PREVIEW = 0
CANNY = 1
BLUR = 2
FEATURES = 3

feature_params = dict(
    maxCorners = 500,
    qualityLevel = 0.2,
    minDistance = 15,
    blockSize = 9
)
image_filter = PREVIEW

source = 0

cap = cv2.VideoCapture(source)

win_name = "Camera Capture"
cv2.namedWindow(win_name, cv2.WINDOW_NORMAL)

Alive = True
while Alive:
    ret, frame = cap.read()
    if not ret:
        break
    frame = cv2.flip(frame,1)
    if image_filter == PREVIEW:
        result = frame
    elif image_filter == CANNY:
        result = cv2.Canny(frame, 80, 150)
    elif image_filter == BLUR:
        result = cv2.blur(frame, (20,20))
    elif image_filter == FEATURES:
        frame_gray = cv2.cvtColor(frame, cv2.COLOR_BGR2GRAY)
        corners = cv2.goodFeaturesToTrack(frame_gray, **feature_params)
        for x , y in np.int32(corners.reshape(-1,2)):
            cv2.circle(frame,(x,y),10,(0,0,255), lineType = 1)
        N = corners.shape[0]
        text = f"Total Corners Found: {N}"
        cv2.putText(frame, text, org = (10,400), fontFace = cv2.FONT_HERSHEY_SIMPLEX, fontScale = 1.5, color = (0,255,255), thickness = 3, lineType = cv2.LINE_AA)
        result = frame

    cv2.imshow(win_name, result)

    key = cv2.waitKey(1)

    if key == ord("q") or key == ord("Q") or key == 27:
        Alive = False
    elif key == ord("p") or key == ord("P"):
        image_filter = PREVIEW
    elif key == ord("c") or key == ord("C"):
        image_filter = CANNY
    elif key == ord("b") or key == ord("B"):
        image_filter = BLUR
    elif key == ord("f") or key == ord("F"):
        image_filter = FEATURES

cap.release()
cv2.destroyWindow(win_name)
```
